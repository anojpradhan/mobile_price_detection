### Proper EDA of the dataset
## Here I will try to find relationship between data features and see which are more related and can be used in mobile to classify the data.

### Import

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Import data

```python
df_mobile_train = pd.read_csv("../data/raw/train.csv")
df_mobile_train.head()
```

```python
df_mobile_train['price_range'].value_counts()
```

Lets understand each feature

```python
df_mobile_train.describe()
```

```python
numeric_cols= df_mobile_train.select_dtypes(include="number").columns
for cols in numeric_cols:
    plt.figure(figsize=(8,4))
    sns.displot(df_mobile_train[cols])
    plt.title(f'Distribution of {cols}')
    plt.show()
```

## From here we can find out that there are 6 binary feature : 
1. Blue
2. four_g
3. three_g
4. dual_sim
5. touch_screen
6. wifi 

```python
binary_cols = ['blue', 'dual_sim', 'four_g', 'three_g','touch_screen', 'wifi']

for cols in binary_cols:
    print(df_mobile_train[cols].value_counts()) 
    print("now by percentage")
    print(df_mobile_train[cols].value_counts(normalize=True) * 100)
```

```python
for cols in numeric_cols:
    if cols != "price_range":
        plt.figure(figsize=(7,4))
        sns.boxplot(data=df_mobile_train, x="price_range", y= cols)
        plt.title(f'{cols} vs Price Range')
        plt.show()
```

```python
for col in numeric_cols:
    if col != 'price_range':
        print(f"\n{col}")
        print(df_mobile_train.groupby('price_range')[col].agg(['mean', 'median', 'min','max']))
```
